## Exercise 5 - Pandas part 2
- Use *gamedata.json* file from data_files directory in this exercise. **Important:** Edit the same DataFrame throughout this whole notebook (changes to DataFrame are applied from each task)!

 1 Read the first 15 rows from the gamedata.json file into the DataFrame. Then read the last 5 rows from the same file into another DataFrame. Combine these two into one DataFrame.

```python
import pandas as pd

df = pd.read_json("data_files/gamedata.json")

df_first_15 = df.head(15)
df_last_5 = df.tail(5)

df_combined = pd.concat([df_first_15, df_last_5], ignore_index=True)

df_combined
```

2 Remove the following columns from the dataframe: internalName, metacriticLink, dealID, storeID, gameID, isOnSale, steamAppID, releaseDate, lastChange, dealRating and thumb. Then keep only rows where savings is 80 % or more and sale price is lower than 3.

```python
columns_to_remove = ['internalName', 'metacriticLink', 'dealID', 'storeID', 'gameID', 'isOnSale', 'steamAppID', 'releaseDate', 'lastChange', 'dealRating', 'thumb']

df_combined = df_combined.drop(columns=columns_to_remove, errors='ignore')

df_combined = df_combined[(df_combined['savings'] >= 80) & (df_combined['salePrice'] < 3)]
df_combined = df_combined.reset_index(drop=True)

df_combined
```

3 Set *steamRatingPercent* to NaN for all rows having *steamRatingCount* below 1000.

```python
df_combined.loc[df_combined['steamRatingCount'] < 1000, 'steamRatingPercent'] = None

df_combined
```

4 Fill all NaN values in *SteamRatingPercent* column by increasing the *metacriticScore* from the same row with 7 %.

```python
score = (df_combined['metacriticScore'] * 1.07).round(1)

df_combined['steamRatingPercent'] = df_combined['steamRatingPercent'].fillna(score)

df_combined
```

 5 Group the data by steamRatingText column so that averages of the following columns will be calculated for each text value: steamRatingPercent, metacriticScore, normalPrice and salePrice. Do not show any other columns in the resulting DataFrame.

```python
columns = ['steamRatingPercent', 'metacriticScore', 'normalPrice', 'salePrice']

df_grouped_input = df_combined[['steamRatingText'] + columns]

df_grouped = df_grouped_input.groupby('steamRatingText', as_index=False).mean()

df_grouped
```
