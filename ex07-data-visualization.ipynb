## Exercise 7 - Data visualization
- Different data sources will be utilized in this exercise.
- **Important**: Remember to adjust the size of each visualized graph using figsize parameter for plot method for better viewing experience!

1 Read the earthquakes.csv file and save its content into a variable. Then use bar chart to visualize magnitude values being 8.2 or over so that date values will be presented in x-axle and magnitude in y-axle.

```python
import pandas as pd

data = pd.read_csv("data_files/earthquakes.csv", delimiter=",")
earthquakes = pd.DataFrame(data)

big_eq = earthquakes[earthquakes['Magnitude'] >= 8.2]

big_eq.plot(kind="bar", x="Date", y="Magnitude", figsize=(16,9), title="Earthquakes with magnitude of at least 8.2", xlabel="Date", ylabel="Magnitude")
```

2 Use the same earthquake data (the unfiltered original data) and visualize the data with line chart so that depth values for the year 2010 are presented. Use date for the x-axle and depth for the y-axle.

```python

earthquakes['Date'] = pd.to_datetime(earthquakes['Date'], format="mixed", utc=True)

year_2010 = earthquakes[earthquakes['Date'].dt.year == 2010]

year_2010.plot(kind="line", x="Date", y="Depth", figsize=(16,9), title="Earthquake depth in 2010", xlabel="Date", ylabel="Depth (km)")
```

3 Read the gamedata.json file to a new variable. Visualize the data with pie chart so that share count will be counted for each Steam rating (in other words how many percentage out of all games have received which rating, for example Very Positive rating).

```python
data = pd.read_json("data_files/gamedata.json")
gamedata = pd.DataFrame(data)

ratings = gamedata["steamRatingText"].value_counts()

ratings.plot.pie(title="Steam ratings", autopct='%1.1f%%', startangle=90, labels=None, figsize=(12,6), pctdistance=1.15).legend(labels=ratings.index, loc="upper left", bbox_to_anchor=(1.0, 1.0))
```

4 Use the same game_data.json as above and create a scatter plot where you compare the relationships for metacriticScore and salePrice.

```python
gamedata.plot(kind="scatter", x="salePrice", y="metacriticScore", figsize=(12,6), title="Price vs. Metacritic Score", xlabel="Sale price", ylabel="Metacritic score")
```

5 Use earthquake data you read earlier (the unfiltered original data) and create a scatter plot for magnitude values using Longitude and Latitude values for x- and y-axles correspondingly. In addition, use colormap for magnitude values.

```python
earthquakes.plot(kind="scatter", x="Longitude", y="Latitude", figsize=(16,9), xlabel="Longitude", ylabel="Latitude", c="Magnitude", colormap="gist_heat_r", vmin=1.0, title="Earthquake locations and magnitudes")
```
